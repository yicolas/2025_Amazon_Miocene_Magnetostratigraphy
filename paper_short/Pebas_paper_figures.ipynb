# Pebas Formation — paper & supplement figures (master notebook)

Regenerates **every figure** in the short manuscript and the supplement from the versioned
scripts in this folder. Set `BASE` (repo root) and `OUT` (figure output dir) in the config cell,
then Run All.

| Figure | What | Generated by |
|---|---|---|
| **Fig 1** | Study area / location map | **external (QGIS)** — see note below |
| **Fig 2** | Master correlation (TAM, SC ↔ La Venta ↔ GPTS GTS2020) | `fig2_master.py` |
| **Fig 3** | Paleomagnetic & rock-magnetic data (+ panel D inclination→paleolatitude) | `fig3_final.py` (imports `fig3_rockmag_dense.py`, `fig3_bottom_candidates.py`) |
| **Fig 4** | Accumulation: age–depth, rate compilation, Sadler | `fig4_combined.py` (imports `fig_sadler.py`) |
| **Fig S1** | Coercivity vs. GRM (greigite screen) | `make_supp_figs.py` |
| **Fig S2** | Reversal in equal-area projection | `make_supp_figs.py` |
| **Fig S3** | Representative Zijderveld diagrams | `make_supp_figs.py` |
| **Fig S4** | Exclusion of alternative GPTS correlations | `fig_altcorr.py` |

*Fig 2 uses `strat_symbols.py` + `symbols/*.svg` (in this folder) and reads data from the repo
(`OSU_uchannel_magnetometer_NRM_data/`, `Pebas-XRF/`, `polarity_synthesis/`,
`age_model_assessment/laventa/`).*

```python
import os
def _find_repo(start):
    d=os.path.abspath(start)
    for _ in range(6):
        if os.path.isdir(os.path.join(d,"OSU_uchannel_magnetometer_NRM_data")) or os.path.basename(d)=="2025_Amazon_Miocene_Magnetostratigraphy":
            return d
        d=os.path.dirname(d)
    return os.path.abspath(os.path.join(start,".."))
# repo root auto-detected from the notebook location; override by setting PEBAS_BASE / PEBAS_OUT
BASE = os.environ.get("PEBAS_BASE") or _find_repo(os.getcwd())
OUT  = os.environ.get("PEBAS_OUT")  or os.path.join(BASE,"figures")
os.environ["PEBAS_BASE"]=BASE; os.environ["PEBAS_OUT"]=OUT
os.makedirs(OUT,exist_ok=True)
from IPython.display import Image, display
print("repo BASE :",BASE); print("figure OUT:",OUT)
```
